```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# ===============================
# Datos ventas productos
# ===============================

ventas = pd.DataFrame({
    "Año":[2018,2019,2020],
    "Producto A":[1500,1700,1600],
    "Producto B":[2000,1900,1800],
    "Producto C":[1300,1400,1500]
})

# GRAFICO 1
ventas.set_index("Año").plot(
    kind="bar",
    figsize=(8,5),
    color=["#1f77b4","#ff7f0e","#2ca02c"]
)
plt.title("Ventas por Producto")
plt.ylabel("Unidades")
plt.show()

# GRAFICO 2
plt.figure(figsize=(8,5))
plt.plot(ventas["Año"],ventas["Producto A"],'o-',label="Producto A")
plt.plot(ventas["Año"],ventas["Producto B"],'s--',label="Producto B")
plt.plot(ventas["Año"],ventas["Producto C"],'^:',label="Producto C")
plt.title("Evolución de Ventas")
plt.legend()
plt.grid(True)
plt.show()

# GRAFICO 3
plt.figure(figsize=(8,5))
for col in ventas.columns[1:]:
    plt.plot(ventas["Año"],ventas[col],marker="o",label=col)

plt.title("Tendencias de Ventas")
plt.legend()
plt.grid(True)
plt.show()

# GRAFICO 4
publicidad = pd.DataFrame({
    "Producto":["A","B","C","D","E"],
    "Publicidad":[20000,50000,30000,150000,25000],
    "Ventas":[100000,250000,200000,500000,120000]
})

plt.figure(figsize=(8,5))
sns.regplot(
    data=publicidad,
    x="Publicidad",
    y="Ventas"
)

for i,row in publicidad.iterrows():
    plt.text(row["Publicidad"],row["Ventas"],row["Producto"])

plt.title("Publicidad vs Ventas")
plt.show()

# GRAFICO 5
deptos = pd.DataFrame({
    "Año":[2018,2019,2020],
    "Dept A":[200000,220000,210000],
    "Dept B":[150000,160000,155000],
    "Dept C":[100000,110000,115000],
    "Dept D":[50000,60000,70000]
})

deptos.set_index("Año").plot(
    kind="bar",
    stacked=True,
    figsize=(8,5)
)
plt.title("Contribución de Departamentos")
plt.ylabel("Ventas")
plt.show()
```
